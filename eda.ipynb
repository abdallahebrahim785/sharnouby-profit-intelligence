## Data Cleaning

```python
# Backup the original dataframes before cleaning
dfs_raw = {name: df.copy() for name, df in dfs.items()}
print(" Raw backup saved")
```

```python
def clean_dataframe(name, df, verbose=True):

    original_shape = df.shape
    df = df.copy()

    # 1) Normalize column names
    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("-", "_")
        .str.lower()
    )

    # 2) Drop fully duplicated rows
    dup_count = df.duplicated().sum()
    df = df.drop_duplicates()

    # 3) Clean text columns
    text_cols = df.select_dtypes(include=["object"]).columns
    for col in text_cols:
        df[col] = df[col].astype(str).str.strip()
        df[col] = df[col].replace({"nan": np.nan, "None": np.nan, "": np.nan})

    # 4) Try numeric conversion for numeric-looking columns
    for col in text_cols:
        sample = df[col].dropna().head(100)
        if len(sample) == 0:
            continue
        try:
            converted = pd.to_numeric(sample, errors="raise")
            if converted.notna().mean() > 0.95:
                df[col] = pd.to_numeric(df[col], errors="coerce")
        except (ValueError, TypeError):
            pass

    if verbose:
        print(f"{name:25s} | {original_shape} -> {df.shape} | "
              f"dropped {dup_count} duplicate rows")

    return df

dfs_clean = {name: clean_dataframe(name, df) for name, df in dfs_raw.items()}
```

```python
def add_date_features(df, date_col, prefix=""):
    """
    Extract time-based features from a date column.
    Creates: year, month, quarter, weekday, week_of_year
    """
    df = df.copy()
    p = f"{prefix}_" if prefix else ""
    df[f"{p}year"]         = df[date_col].dt.year
    df[f"{p}month"]        = df[date_col].dt.month
    df[f"{p}quarter"]      = df[date_col].dt.quarter
    df[f"{p}weekday"]      = df[date_col].dt.day_name()
    df[f"{p}week_of_year"] = df[date_col].dt.isocalendar().week.astype("Int64")
    df[f"{p}ym"]           = df[date_col].dt.to_period("M").astype(str)  # "2024-01"
    return df


# ===== sales_transactions =====
sales = dfs_clean["sales_transactions"]
sales = add_date_features(sales, "order_date")
dfs_clean["sales_transactions"] = sales
print("✅ sales_transactions: date features added")

# ===== production_orders =====
po = dfs_clean["production_orders"]
po = add_date_features(po, "start_date")
dfs_clean["production_orders"] = po
print("✅ production_orders: date features added")

# ===== quality =====
q = dfs_clean["quality"]
q = add_date_features(q, "inspection_date")
dfs_clean["quality"] = q
print("✅ quality: date features added")

# ===== purchase_orders =====
purch = dfs_clean["purchase_orders"]
purch = add_date_features(purch, "order_date")
dfs_clean["purchase_orders"] = purch
print("✅ purchase_orders: date features added")
```

```python
sales = dfs_clean["sales_transactions"]
products = dfs_clean["product_master"]
customers = dfs_clean["customers"]
locations = dfs_clean["locations"]

def check_join(left_df, left_key, right_df, right_key, label):
    left_vals = set(left_df[left_key].dropna().unique())
    right_vals = set(right_df[right_key].dropna().unique())
    common = left_vals & right_vals
    missing = left_vals - right_vals
    pct = len(missing) / len(left_vals) * 100 if left_vals else 0
    status = "✅" if len(missing) == 0 else "⚠️"
    print(f"{status} {label}")
    print(f"   Left: {len(left_vals):,} | Right: {len(right_vals):,} "
          f"| Common: {len(common):,} | Missing: {len(missing):,} ({pct:.2f}%)\n")

check_join(sales, "product_id", products, "product_id", "sales → product_master")
check_join(sales, "customer_id", customers, "customer_id", "sales → customers")
check_join(sales, "location_id", locations, "location_id", "sales → locations")
```

```python
sales = dfs_clean["sales_transactions"].copy()
products = dfs_clean["product_master"].copy()
customers = dfs_clean["customers"].copy()
locations = dfs_clean["locations"].copy()

# Join product info
prod_cols = [c for c in products.columns if c not in sales.columns and c != "product_id"]
sales_master = sales.merge(
    products[["product_id"] + prod_cols],
    on="product_id", how="left", suffixes=("", "_prod")
)
print(f"After products: {sales_master.shape}")

# Join customer info
cust_cols = [c for c in customers.columns if c not in sales_master.columns and c != "customer_id"]
sales_master = sales_master.merge(
    customers[["customer_id"] + cust_cols],
    on="customer_id", how="left", suffixes=("", "_cust")
)
print(f"After customers: {sales_master.shape}")

# Join location info
loc_cols = [c for c in locations.columns if c not in sales_master.columns and c != "location_id"]
sales_master = sales_master.merge(
    locations[["location_id"] + loc_cols],
    on="location_id", how="left", suffixes=("", "_loc")
)
print(f"After locations: {sales_master.shape}")

dfs_clean["sales_master"] = sales_master
print(f"\n✅ sales_master created: {sales_master.shape}")
```

```python
sales_master = dfs_clean["sales_master"]

# Product-level aggregation
product_summary = (
    sales_master
    .groupby(["product_id", "product_name", "product_family"])
    .agg(
        total_quantity    = ("quantity", "sum"),
        total_net_sales   = ("net_sales_egp", "sum"),
        total_cost        = ("estimated_cogs_egp", "sum"),
        total_profit      = ("gross_profit_egp", "sum"),
        avg_margin_pct    = ("gross_margin_pct", "mean"),
        avg_discount_pct  = ("discount_pct", "mean"),
        n_transactions    = ("invoice_id", "nunique"),
    )
    .reset_index()
)

```

```python
# Compute profit margin for each product
product_summary["profit_margin_pct"] = (
    product_summary["total_profit"] / product_summary["total_net_sales"] * 100
).round(2)

print(f"📦 Total unique products: {len(product_summary):,}")
print(f"📊 Total profit: {product_summary['total_profit'].sum():,.0f} EGP")
print(f"📊 Total net sales: {product_summary['total_net_sales'].sum():,.0f} EGP")
print(f"\n✅ product_summary created: {product_summary.shape}")
```

# Scenario 2:
## Customer Segmentation (RFM)

```python
sales_master = dfs_clean["sales_master"]

# Reference date = day after the last order in the dataset
reference_date = sales_master["order_date"].max() + pd.Timedelta(days=1)
print(f"📅 Reference date: {reference_date}")

# RFM calculation per customer
rfm = (
    sales_master
    .groupby("customer_id")
    .agg(
        recency_days   = ("order_date", lambda x: (reference_date - x.max()).days),
        frequency      = ("invoice_id", "nunique"),
        monetary       = ("net_sales_egp", "sum"),
    )
    .reset_index()
)

print(f"\n✅ RFM computed for {len(rfm):,} customers")
print(f"\nSample:")
print(rfm.head(10).to_string(index=False))

print(f"\nDistribution:")
print(rfm[["recency_days", "frequency", "monetary"]].describe().T.to_string())
```

### Score each customer 1-5 on R, F, M

```python

rfm["R_score"] = pd.qcut(rfm["recency_days"], 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm["F_score"] = pd.qcut(rfm["frequency"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm["M_score"] = pd.qcut(rfm["monetary"], 5, labels=[1, 2, 3, 4, 5]).astype(int)

# Combined RFM score
rfm["RFM_score"] = rfm["R_score"].astype(str) + rfm["F_score"].astype(str) + rfm["M_score"].astype(str)

# Weighted total score (simple sum)
rfm["RFM_total"] = rfm["R_score"] + rfm["F_score"] + rfm["M_score"]

print("✅ RFM scores assigned")
print(f"\nSample with scores:")
print(rfm.head(10).to_string(index=False))
```

```python
def segment_customer(row):
    r, f, m = row["R_score"], row["F_score"], row["M_score"]
    
    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"
    elif r >= 4 and f >= 3 and m >= 3:
        return "Loyal"
    elif r >= 4 and f <= 2:
        return "New Customers"
    elif r >= 3 and f >= 2 and m >= 2:
        return "Potential Loyalists"
    elif r <= 2 and f >= 3 and m >= 3:
        return "At Risk"
    elif r <= 2 and f <= 2 and m <= 2:
        return "Lost"
    elif r <= 2 and f >= 4 and m >= 4:
        return "Can't Lose Them"
    else:
        return "Need Attention"

rfm["segment"] = rfm.apply(segment_customer, axis=1)

```

```python
# Summary
segment_summary = (
    rfm.groupby("segment")
    .agg(
        n_customers   = ("customer_id", "count"),
        total_revenue = ("monetary", "sum"),
        avg_recency   = ("recency_days", "mean"),
        avg_frequency = ("frequency", "mean"),
        avg_monetary  = ("monetary", "mean"),
    )
    .round(2)
    .sort_values("total_revenue", ascending=False)
    .reset_index()
)

segment_summary["revenue_pct"] = (
    segment_summary["total_revenue"] / segment_summary["total_revenue"].sum() * 100
).round(2)
segment_summary["customer_pct"] = (
    segment_summary["n_customers"] / segment_summary["n_customers"].sum() * 100
).round(2)

print("🎯 Customer Segments:\n")
print(segment_summary.to_string(index=False))

dfs_clean["rfm"] = rfm
```

```python
print("📊 Segment breakdown — % of customers vs % of revenue:\n")

print(segment_summary[["segment", "n_customers", "customer_pct", 
                       "total_revenue", "revenue_pct", 
                       "avg_monetary", "avg_frequency"]].to_string(index=False))
```

```python
print(" Average revenue per customer by segment:\n")

segment_summary["revenue_per_customer"] = (
    segment_summary["total_revenue"] / segment_summary["n_customers"]
).round(0)

print(
    segment_summary[["segment", "n_customers", "revenue_per_customer", "avg_frequency"]]
    .sort_values("revenue_per_customer", ascending=False)
    .to_string(index=False)
)
```

```python
print(" One-time buyers analysis:\n")

one_time = rfm[rfm["frequency"] == 1]
repeat = rfm[rfm["frequency"] > 1]

print(f"Total customers: {len(rfm):,}")
print(f"One-time buyers: {len(one_time):,} ({len(one_time)/len(rfm)*100:.2f}%)")
print(f"Repeat buyers:   {len(repeat):,} ({len(repeat)/len(rfm)*100:.2f}%)")

print(f"\nRevenue contribution:")
print(f"  One-time: {one_time['monetary'].sum():,.0f} EGP ({one_time['monetary'].sum()/rfm['monetary'].sum()*100:.2f}%)")
print(f"  Repeat:   {repeat['monetary'].sum():,.0f} EGP ({repeat['monetary'].sum()/rfm['monetary'].sum()*100:.2f}%)")

print(f"\nAverage value per customer:")
print(f"  One-time: {one_time['monetary'].mean():,.0f} EGP")
print(f"  Repeat:   {repeat['monetary'].mean():,.0f} EGP")
```

```python
sales_master = dfs_clean["sales_master"]

print(" Available Channels:\n")
print(sales_master["channel"].value_counts().to_string())

print("\n Available Locations:\n")
print(sales_master["location_name"].value_counts().to_string())

print("\n Cross-tab: Channel × Location\n")
print(pd.crosstab(
    sales_master["location_name"],
    sales_master["channel"],
    margins=True
).to_string())
```

```python
print(" Channel Performance:\n")

channel_summary = (
    sales_master
    .groupby("channel")
    .agg(
        n_transactions = ("invoice_id", "nunique"),
        n_customers    = ("customer_id", "nunique"),
        total_quantity = ("quantity", "sum"),
        total_net_sales= ("net_sales_egp", "sum"),
        total_profit   = ("gross_profit_egp", "sum"),
        avg_margin     = ("gross_margin_pct", "mean"),
        avg_discount   = ("discount_pct", "mean"),
    )
    .round(2)
    .sort_values("total_net_sales", ascending=False)
    .reset_index()
)

channel_summary["revenue_pct"] = (
    channel_summary["total_net_sales"] / channel_summary["total_net_sales"].sum() * 100
).round(2)
channel_summary["profit_pct"] = (
    channel_summary["total_profit"] / channel_summary["total_profit"].sum() * 100
).round(2)
channel_summary["avg_order_value"] = (
    channel_summary["total_net_sales"] / channel_summary["n_transactions"]
).round(0)

print(channel_summary.to_string(index=False))
```

```python
print(" Location Performance:\n")

location_summary = (
    sales_master
    .groupby(["location_id", "location_name"])
    .agg(
        n_transactions = ("invoice_id", "nunique"),
        n_customers    = ("customer_id", "nunique"),
        total_quantity = ("quantity", "sum"),
        total_net_sales= ("net_sales_egp", "sum"),
        total_profit   = ("gross_profit_egp", "sum"),
        avg_margin     = ("gross_margin_pct", "mean"),
    )
    .round(2)
    .sort_values("total_net_sales", ascending=False)
    .reset_index()
)

location_summary["revenue_pct"] = (
    location_summary["total_net_sales"] / location_summary["total_net_sales"].sum() * 100
).round(2)

print(location_summary.to_string(index=False))
```

```python
print(" Governorate Performance (Top 15):\n")

gov_summary = (
    sales_master
    .groupby("customer_governorate")
    .agg(
        n_customers    = ("customer_id", "nunique"),
        n_transactions = ("invoice_id", "nunique"),
        total_net_sales= ("net_sales_egp", "sum"),
        total_profit   = ("gross_profit_egp", "sum"),
        avg_margin     = ("gross_margin_pct", "mean"),
    )
    .round(2)
    .sort_values("total_net_sales", ascending=False)
    .head(15)
    .reset_index()
)

gov_summary["revenue_pct"] = (
    gov_summary["total_net_sales"] / sales_master["net_sales_egp"].sum() * 100
).round(2)

print(gov_summary.to_string(index=False))
```

```python
print(" Channel × Location — Revenue Matrix:\n")

matrix_revenue = sales_master.pivot_table(
    index="location_name",
    columns="channel",
    values="net_sales_egp",
    aggfunc="sum",
    fill_value=0
).round(0)

print(matrix_revenue.to_string())

print("\n📊 Channel × Location — Profit Margin % Matrix:\n")

matrix_margin = sales_master.pivot_table(
    index="location_name",
    columns="channel",
    values="gross_margin_pct",
    aggfunc="mean",
    fill_value=0
).round(2)

print(matrix_margin.to_string())
```

```python
print(" Channel × Customer Type:\n")

channel_customer = pd.crosstab(
    sales_master["channel"],
    sales_master["customer_type"],
    values=sales_master["net_sales_egp"],
    aggfunc="sum",
    margins=True
).round(0)

print(channel_customer.to_string())

print("\n📊 Average Order Value by Channel × Customer Type:\n")

aov_matrix = sales_master.pivot_table(
    index="channel",
    columns="customer_type",
    values="net_sales_egp",
    aggfunc="mean"
).round(0)

print(aov_matrix.to_string())
```
